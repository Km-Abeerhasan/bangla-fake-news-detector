## 1. Input Claim & Empirical Evidence Pool
Provide a simulated real-world claim alongside audited snapshot passages.

```python
# Real-world simulated Bengali news claim
claim_text = "তিস্তা নদীর বাঁধ খুলে দেওয়ায় নিম্নাঞ্চল প্লাবিত হয়েছে।"
claim_time = datetime(2026, 8, 12)

# Retrieved empirical snapshot passages with source tiers
evidence_pool = {
    "doc101": {
        "text": "পানি উন্নয়ন বোর্ড নিশ্চিত করেছে তিস্তা ব্যারেজের সবকটি জলকপাট খুলে দেওয়া হয়েছে।",
        "publisher": "Prothom Alo",
        "tier": 3,
        "url": "https://prothomalo.com/news/101",
        "t_pub": datetime(2026, 8, 10)
    },
    "doc102": {
        "text": "উজানের ঢলে তিস্তা অববাহিকার চরাঞ্চল ও নিম্নাঞ্চল প্লাবিত হয়েছে বলে জানিয়েছে প্রশাসন।",
        "publisher": "Daily Star",
        "tier": 3,
        "url": "https://thedailystar.net/news/102",
        "t_pub": datetime(2026, 8, 10)
    }
}

print(f"Claim: {claim_text}")
print(f"Evidence Passages Loaded: {len(evidence_pool)}")
```

## 2. Execute End-to-End BiN Pipeline (S1 to S8)

```python
pipeline = BiNPipeline(tau_abstain=0.60)
report = pipeline.run_pipeline(
    raw_text=claim_text,
    evidence_pool=evidence_pool,
    t_claim=claim_time,
    claim_id="DEMO-CLM-001"
)

print(f"Pipeline Status: {report['gate_status']}")
print(f"Verdict Label:   {report['verdict']['label']}")
print(f"Rule Fired:      {report['verdict']['rule_fired']}")
print(f"Citation Audit:  {'PASSED' if report['citation_audit']['passed'] else 'FAILED'}")
```
